# Data Cleaning

Import libraries and dataset

```python
# import required 
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('Registration_Data_2025_CPUT_Students.csv')

# Create a copy to work on, keeping the original safe
df_clean = df.copy()
```

 handle duplicates 


```python
# Remove duplicates based on Identifier
if 'Identifier' in df_clean.columns:
    df_clean.drop_duplicates(subset=['Identifier'], keep='first', inplace=True)
else:
    df_clean.drop_duplicates(inplace=True)
```

Handle missiwng values

```python
print(df.isnull().sum())
```

```python
# fill the missing values with the mean or the median
num_cols = ['School Quintile', 'Matric Date', 'Age at Year_Beginning'] 


for col in num_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
# for categorical columns fill with "Unknown"

cat_cols = df_clean.select_dtypes(include=['object']).columns

for col in cat_cols:
    df_clean[col] = df_clean[col].fillna('Unknown')
```

Clean Text Data (Standardization)

```python
# clean text data

# Strip whitespace and standardize to Title Case for all object columns
for col in cat_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip().str.title()

# Specific fix for Gender (if it contains 'MALE', 'FEMALE', 'Male', 'Female')
if 'Gender' in df_clean.columns:
    df_clean['Gender'] = df_clean['Gender'].replace({
        'Male': 'M', 
        'Female': 'F', 
        'MALE': 'M', 
        'FEMALE': 'F'
    })
#change columns, Not entering to Not registered and the rest to registered
if 'Ften Status This Year' in df_clean.columns:
    df_clean['Ften Status This Year'] = df_clean['Ften Status This Year'].replace({
        'Non Entering' : 'Not Registered',
        'Entering':'Registered',
        'First Time Entering':'Registered',
        'Transfer':'Registered'    
    })
```

Verification

```python
print(df_clean.info())
print(df_clean.head())
```

Save the cleaned data

```python
# Save the cleaned dataframe as a CSV file
df_clean.to_csv('Registration_Data_Cleaned.csv', index=False)

print("File successfully saved as Registration_Data_Cleaned.csv")
```
